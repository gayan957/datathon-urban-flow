```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
taxi = pd.read_csv("../data/processed/df_integrated.csv")
```

```python
train, temp = train_test_split(taxi, test_size=0.3, random_state=42)
val, test = train_test_split(temp, test_size=0.5, random_state=42)

print(f"Train: {len(train)} ({len(train)/len(taxi)*100:.1f}%)")
print(f"Val:   {len(val)} ({len(val)/len(taxi)*100:.1f}%)")
print(f"Test:  {len(test)} ({len(test)/len(taxi)*100:.1f}%)")

train.to_csv('../data/splits/train.csv', index=False)
val.to_csv('../data/splits/val.csv', index=False)
test.to_csv('../data/splits/test.csv', index=False)
```
